```python
import pandas as pd
import re
```

## DataLoad

```python
train_df = pd.read_csv('../Data/train.csv')
dev_df = pd.read_csv('../Data/dev.csv')
test_df = pd.read_csv('../Data/test.csv')
```

## 전처리 통합 코드

전략1. HTML 찌꺼기 제거해보기 <br>
전략2. 괄호 한의 행동 지시문 제거해보기 (모델이 행동보다 텍스트에 조금 더 집중 할 수 있도록) <br>
전략3. 구어체 추임새 제거해보기 (문장 맨 앞이나 쉼표와 함께 쓰이는 무의미한 감탄사들)

```python
def clean_dialogue(text):
    if not isinstance(text, str):
        return text
    
    # 1. HTML 찌꺼기 제거
    text = text.replace('<br', '\n')
    text = text.replace('&quot;', '"')

    # 2. 괄호 안의 행동 지시문 제거
    text = re.sub(r'\([^)]*\)', '', text)

    # 3. 구어체 추임새 제거
    fillers = [
        '아, ', '어, ', '음, ', '그니까, ', '근데, ', '그냥, ', 
        '막, ', '저기요, ', '있잖아, ', '에휴, ', '와, ', '오, '
    ]
    for filler in fillers:
        text = text.replace(filler, '')

    # 다중 공백도 함께 정리
    text = re.sub(r' +', ' ', text)

    return text.strip()
```

## Data 에 전처리 요소 적용하기

```python
for df in [train_df, dev_df, test_df]:
    df['dialogue'] = df['dialogue'].apply(clean_dialogue)
    # 대화문 앞에 dialogue 와 같은 명시적 태그를 달아서 성능 향상에 기여
    df['dialogue'] = "[Dialogue]\n" + df['dialogue']

print("전처리 샘플 확인 해보기 : ", train_df['dialogue'].iloc[0][:150] + "...")
```

## Save

```python
train_df.to_csv('../Data/train_processed.csv', index=False, encoding='utf-8-sig')
dev_df.to_csv('../Data/dev_processed.csv', index=False, encoding='utf-8-sig')
test_df.to_csv('../Data/test_processed.csv', index=False, encoding='utf-8-sig')
```
